```python
# Assess weights to ingredients
# Overabundance of "safe" stuff caused class imbalance
from pathlib import Path
import pandas as pd

DATA = Path("dataset.csv")
OUT = Path("dataset_weighted.csv")

df = pd.read_csv(DATA)

def add_weights(df, label):
    counts = df[label].value_counts()
    total = len(df)
    w_map = {cls: total / cnt for cls, cnt in counts.items()}
    return df[label].map(w_map)

for col in ["celiac", "vegan", "nut_allergy"]:
    df[f"{col}_weight"] = add_weights(df, col)
    minor_mask = df[col].isin(["definitely unsafe", "potentially unsafe"])
    df.loc[minor_mask, f"{col}_weight"] *= 2

df.to_csv(OUT, index=False)
```

```python
# TF-IDF
import pandas as pd, numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

DF = pd.read_csv("dataset_weighted.csv")

# Common vectoriser, n-gram
VEC = TfidfVectorizer(analyzer="char", ngram_range=(2,5), min_df=3)

def train_one_task(col):
    print(f"\n=== {col.upper()} ===")
    X = DF["ingredient_name"].astype(str)
    y = DF[col]
    w = DF[f"{col}_weight"]

    # 70 / 15 / 15 split, stratified
    X_train, X_temp, y_train, y_temp, w_train, w_temp = train_test_split(
        X, y, w, test_size=0.30, random_state=42, stratify=None
    )

    X_val, X_test, y_val, y_test, w_val, w_test = train_test_split(
        X_temp, y_temp, w_temp, test_size=0.50, random_state=42, stratify=y_temp)

    # Fit vectoriser on TRAIN ONLY
    X_train_vec = VEC.fit_transform(X_train)
    X_val_vec   = VEC.transform(X_val)
    X_test_vec  = VEC.transform(X_test)

    # Class-weighting via sample_weight
    clf = LogisticRegression(max_iter=400, n_jobs=-1)
    clf.fit(X_train_vec, y_train, sample_weight=w_train)

    # Validation
    print("\nValidation")
    print(classification_report(y_val, clf.predict(X_val_vec)))
    # Testing
    print("Test")
    print(classification_report(y_test, clf.predict(X_test_vec)))

    return clf

# Train all three tasks
models = {col: train_one_task(col) for col in ["celiac","vegan","nut_allergy"]}
```

```python
import random, gc, numpy as np, torch, torch.nn.functional as F
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from torch.optim import AdamW
from torch.utils.data import Dataset, DataLoader

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)

df = pd.read_csv("dataset_weighted.csv")
TEXT_COL = "ingredient_name"
LABEL_COLS = ["celiac", "vegan", "nut_allergy"]
WEIGHT_COL = [c + "_weight" for c in LABEL_COLS]
label2id = {"safe": 0, "potentially unsafe": 1, "definitely unsafe": 2}
df[LABEL_COLS] = df[LABEL_COLS].replace(label2id)

train_df, tmp_df = train_test_split(df, test_size=0.30, random_state=SEED)
val_df, test_df = train_test_split(tmp_df, test_size=0.50, random_state=SEED)

# OCR spelling normalization
def normalize_text(s):
    return s.lower().replace("sut", "süt").replace("yogurt", "yoğurt").replace("peynir", "peynir")

tok = AutoTokenizer.from_pretrained("bert-base-multilingual-cased")

class IngredientDS(Dataset):
    def __init__(self, pdf, y_col, w_col):
        clean_texts = pdf[TEXT_COL].apply(normalize_text).tolist()
        self.enc = tok(clean_texts, truncation=True, padding=True)
        self.labels = pdf[y_col].astype(int).values
        self.weights = pdf[w_col].astype(float).values
    def __len__(self): return len(self.labels)
    def __getitem__(self, idx):
        item = {k: torch.tensor(v[idx]) for k, v in self.enc.items()}
        item["labels"] = torch.tensor(self.labels[idx])
        item["weights"] = torch.tensor(self.weights[idx], dtype=torch.float)
        return item

def label_smoothed_nll_loss(logits, target, smoothing=0.1):
    log_probs = F.log_softmax(logits, dim=-1)
    n_classes = logits.size(-1)
    with torch.no_grad():
        true_dist = torch.zeros_like(log_probs)
        true_dist.fill_(smoothing / (n_classes - 1))
        true_dist.scatter_(1, target.unsqueeze(1), 1.0 - smoothing)
    return -(true_dist * log_probs).sum(dim=-1)

def run_epoch(loader, model, optim=None):
    is_train = optim is not None
    all_pred, all_true, losses = [], [], []
    model.train() if is_train else model.eval()
    torch.set_grad_enabled(is_train)

    for batch in loader:
        batch = {k: v.to(DEVICE) for k, v in batch.items()}
        out = model(input_ids=batch["input_ids"],
                    attention_mask=batch["attention_mask"],
                    token_type_ids=batch.get("token_type_ids", None))
        loss_raw = label_smoothed_nll_loss(out.logits, batch["labels"])
        loss = (loss_raw * batch["weights"]).mean()
        if is_train:
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optim.step(); optim.zero_grad()
        losses.append(loss.item())
        preds = out.logits.argmax(-1).detach().cpu().numpy()
        all_pred.extend(preds)
        all_true.extend(batch["labels"].cpu().numpy())

    acc = accuracy_score(all_true, all_pred)
    pr, rc, f1, _ = precision_recall_fscore_support(all_true, all_pred, average="macro", zero_division=0)
    return np.mean(losses), acc, pr, rc, f1

EPOCHS, BATCH = 20, 16
results = {}

for i, lbl in enumerate(LABEL_COLS):
    print(f"\n🟣  {lbl.upper()}  =============================")
    train_ds = IngredientDS(train_df, lbl, WEIGHT_COL[i])
    val_ds = IngredientDS(val_df, lbl, WEIGHT_COL[i])
    test_ds = IngredientDS(test_df, lbl, WEIGHT_COL[i])

    train_loader = DataLoader(train_ds, batch_size=BATCH, shuffle=True)
    val_loader = DataLoader(val_ds, batch_size=BATCH)
    test_loader = DataLoader(test_ds, batch_size=BATCH)

    model = AutoModelForSequenceClassification.from_pretrained(
        "bert-base-multilingual-cased",
        num_labels=3,
        hidden_dropout_prob=0.3,
        attention_probs_dropout_prob=0.2
    ).to(DEVICE)
    optim = AdamW(model.parameters(), lr=2e-5)

    best_f1, best_state, stop_count = 0, None, 0
    for ep in range(1, EPOCHS + 1):
        tr = run_epoch(train_loader, model, optim)
        vl = run_epoch(val_loader, model)
        print(f"Epoch {ep:02d} | train_loss {tr[0]:.4f} | val_loss {vl[0]:.4f} | val_acc {vl[1]:.3f} | val_f1 {vl[4]:.3f}")
        if vl[4] > best_f1:
            best_f1, best_state = vl[4], {k: v.cpu() for k, v in model.state_dict().items()}
            stop_count = 0
        else:
            stop_count += 1
            if stop_count >= 3:
                print("Early stopping triggered.")
                break

    model.load_state_dict(best_state)
    tst = run_epoch(test_loader, model)
    results[lbl] = tst
    print(f"TEST  ->  loss {tst[0]:.4f} | acc {tst[1]:.3f} | f1 {tst[4]:.3f}")
    torch.save(model.state_dict(), f"/content/{lbl}_bert_model.pt")
    del model, optim; torch.cuda.empty_cache(); gc.collect()

print("\n===========  FINAL TEST METRICS  ===========")
for lbl, (loss, acc, pr, rc, f1) in results.items():
    print(f"{lbl.upper():10s}  loss {loss:.4f} | acc {acc:.3f} | P {pr:.3f}  R {rc:.3f}  F1 {f1:.3f}")
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

def load_model(label_name):
    model = AutoModelForSequenceClassification.from_pretrained(
        "bert-base-multilingual-cased", num_labels=3)
    model.load_state_dict(torch.load(f"/content/{label_name}_bert_model.pt", map_location=DEVICE))
    model.to(DEVICE).eval()
    tokenizer = AutoTokenizer.from_pretrained("bert-base-multilingual-cased")
    return model, tokenizer

def predict_one(text, label_name):
    model, tokenizer = load_model(label_name)
    enc = tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(DEVICE)
    with torch.no_grad():
        logits = model(**enc).logits
        pred = logits.argmax(-1).item()
    return {0: "safe", 1: "potentially unsafe", 2: "unsafe"}[pred]

def predict_many(text_list, label_name):
    model, tokenizer = load_model(label_name)
    enc = tokenizer(text_list, return_tensors="pt", truncation=True, padding=True).to(DEVICE)
    with torch.no_grad():
        logits = model(**enc).logits
        preds = logits.argmax(-1).cpu().numpy()
    id2label = {0: "safe", 1: "potentially unsafe", 2: "unsafe"}
    return [id2label[p] for p in preds]
```
